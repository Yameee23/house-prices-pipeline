```python
# House Prices - Pipeline de nettoyage
```

## 1. Imports & Chargement

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
df=pd.read_csv("house_prices.csv")
```

## 2. Exploration


```python
df.info()

df.isnull().sum()
```

## 3. Nettoyage

```python
df.drop(["Society", "Description", "facing","overlooking","Amount(in rupees)","Dimensions","Plot Area", "Title"], axis=1, inplace=True)
```

```python
 #Suppression des lignes avec valeurs manquantes critiques
df.dropna(subset=["Transaction", "Bathroom"], inplace=True)
```

```python
#permet de remplir les colonnes texte par "non renseigne
df.fillna({"Balcony": "Non renseigné","Furnishing": "Non renseigné","Status" : "Non renseigné","Floor" : "Non renseigné","Car Parking" : "Non renseigné" , "Ownership" : " Non renseigné"}, inplace=True)
```

```python
df["Bathroom"] = pd.to_numeric(df["Bathroom"], errors="coerce")
df["Balcony"] = pd.to_numeric(df["Balcony"], errors="coerce")
```

```python
df["Super Area"] = pd.to_numeric(df["Super Area"].str.replace(" sqft", ""), errors="coerce")
df["Carpet Area"] = pd.to_numeric(df["Carpet Area"].str.replace(" sqft", ""), errors="coerce")
```

## 4. Encodage

```python
df["Furnishing"] = df["Furnishing"].map({"Unfurnished" : 0, "Semi-Furnished" : 1 , "Furnished" : 2 })
df["Status"] = df["Status"].map({"Ready to Move" : 0, "Non renseigné" : 1})
df["Transaction"] = df["Transaction"].map({"Resale" : 0, "New Property" : 1, "Other" : 2 , "Rent/Lease" : 3})
df["Ownership"] = df["Ownership"].map({"Freehold" : 0, "Non renseigné" : 1, "Leasehold" : 2 , "Co-operative Society" : 3, "Power Of Attorney" :4})
```

```python
le = LabelEncoder()
df['location'] = le.fit_transform((df['location']))
```

## 5. Normalisation

```python
Bathroom_min = df["Bathroom"].min()
Bathroom_max= df["Bathroom"].max()
Price_min = df["Price (in rupees)"].min()
Price_max= df["Price (in rupees)"].max()
Carpet_min = df["Carpet Area"].min()
Carpet_max = df["Carpet Area"].max()
Super_Area_min = df["Super Area"].min()
Super_Area_max = df["Super Area"].max()
Balcony_min = df["Balcony"].min()
Balcony_max= df["Balcony"].max()
```

```python
df["Bathroom"] = (df["Bathroom"] - Bathroom_min ) /( Bathroom_max-Bathroom_min)
df["Price (in rupees)"] = (df["Price (in rupees)"] - Price_min ) /( Price_max - Price_min)
df["Carpet Area"] = (df["Carpet Area"] - Carpet_min) / (Carpet_max - Carpet_min)
df["Super Area"] = (df["Super Area"] - Super_Area_min) / (Super_Area_max - Super_Area_min)
df["Balcony"] = (df["Balcony"] - Balcony_min ) /( Balcony_max-Balcony_min)
```

```python
df.to_csv("house_prices_final.csv", index=False)
```
